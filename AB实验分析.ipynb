```python
import pandas as pd
from nltk.sem.chat80 import sql_query
from sqlalchemy import create_engine
from statsmodels.stats.proportion import proportions_ztest
#创建连接
engine = create_engine('mysql+pymysql://root:147369@localhost:3306/itcast')
df = pd.read_sql("select * from stg_ab_test_assignment", engine)
print("连接成功！已读取数据：",len(df))

#1.统计两组的总人数（从刚刚读入的df自动获取）
n_control = len(df[df['test_group'] == 'Control'])
n_treatment = len(df[df['test_group'] == 'Treatment'])

#2.模拟实验结果（假设7天后）
#假设对照组（没券）有23人下单，实验组（有券）有41人下单
successes = [23,41]
nobs = [n_control,n_treatment]

#执行z检验
z_stat,p_val = proportions_ztest(successes,nobs)

print(f"--- A/B测试报告 ---")
print(f"对照组转化率：{successes[0]/nobs[0]:.2%}")
print(f"实验组转化率：{successes[1]/nobs[1]:.2%}")
print(f"P值 (P-Value):{p_val:.4f}")

#给出最后结论
if p_val < 0.05:
    print("\n结论：结果在统计学上是显著的!")
    print("优惠券策略缺失提升了复购，建议在全体潜在流失客户中推广。")
else:
    print("\n结论：结果不显著。")
    print("虽然实验组看起来高一点，但在统计学上无法排除是'运气'导致的。建议重新设计优惠券额度或加大样本量。")
```
